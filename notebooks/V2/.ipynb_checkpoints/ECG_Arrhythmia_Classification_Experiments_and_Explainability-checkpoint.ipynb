# Improved 1D CNN for ECG Arrhythmia Classification

**B.Tech Project:** Explainable Deep Learning Framework for Automated ECG Arrhythmia Detection and Classification

This notebook starts from the baseline 1D CNN and adds six improvements **one at a time**, so the effect of each one can be seen in a results table.

| Step | Change added | Problem in the baseline it targets |
|---|---|---|
| B0 | Baseline (numbers from the earlier run) | Accuracy 94.41%, macro-F1 76.29%, F precision only 26.02% |
| E1 | Choose the best epoch by **validation macro-F1** | The baseline picked its epoch on a validation loss dominated by class N |
| E2 | + stable training: lower learning rate, gradient clipping, faster batch-norm statistics | Validation accuracy jumped between 14% and 95% from epoch to epoch |
| E3 | + softer class weights (square root) | One F beat counted as much as about 114 N beats, so the model over-predicted F and S |
| E4 | + light augmentation of training beats | Rare classes have few training beats (F has about 513) |
| E5 | + deeper network with residual blocks | Training accuracy was only 90.93% and still rising when training stopped |
| Final | Best of E1 to E5 (chosen on validation) + tuned decision thresholds | F and S were predicted too often |

**Rules followed:** same seed and same train / validation / test split as the baseline. The test set is never used to pick an epoch, a model or a threshold. Only the validation set is used for that.

**Dataset:** Kaggle *ECG Heartbeat Categorization Dataset* (MIT-BIH part), 187 samples per beat, 125 Hz, classes N, S, V, F, Q. The split is by beat, not by patient, so these are intra-patient scores.

## 1. Setup

```python
import os, random, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (classification_report, confusion_matrix, f1_score,
                             precision_recall_fscore_support, accuracy_score)
warnings.filterwarnings('ignore')

SEED = 42
def reset_seeds():
    random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)
reset_seeds()

CLASS_NAMES = ['N', 'S', 'V', 'F', 'Q']          # label 0, 1, 2, 3, 4
CLASS_FULL = {'N': 'Normal', 'S': 'Supraventricular ectopic', 'V': 'Ventricular ectopic',
              'F': 'Fusion', 'Q': 'Unknown / paced'}
FS = 125             # sampling frequency in Hz
BATCH_SIZE = 128
EPOCHS = 30          # maximum epochs per experiment
PATIENCE = 8         # stop if validation macro-F1 has not improved for this many epochs
RUN_IDS = ['E1', 'E2', 'E3', 'E4', 'E5']   # remove an id to skip that experiment

# Scores of the baseline notebook, copied from its printed output (earlier run)
BASELINE = {'acc': 0.9441, 'macro_f1': 0.7629, 'rec_S': 0.6691, 'rec_F': 0.8642,
            'prec_S': 0.5586, 'prec_F': 0.2602}

try:
    gpus = tf.config.list_physical_devices('GPU')
except AttributeError:                      # older TensorFlow
    gpus = tf.config.experimental.list_physical_devices('GPU')
print('TensorFlow', tf.__version__, '| GPU found:', bool(gpus))
if not gpus:
    RUN_IDS, EPOCHS = ['E4'], 15
    print('WARNING: no GPU. Running all five experiments on CPU would take more than an hour,')
    print('so only E4 (all changes except the deeper network) will run, for 15 epochs.')
    print('Turn the GPU on (Settings -> Accelerator) and run again for the full comparison.')

FIG_DIR = 'figures'
os.makedirs(FIG_DIR, exist_ok=True)
def savefig(name):
    plt.savefig(os.path.join(FIG_DIR, name), dpi=200, bbox_inches='tight')
```

## 2. Load the data

```python
def find_data_dir():
    for root in [os.environ.get('ECG_DATA_DIR', ''), '/kaggle/input', '.']:
        if root and os.path.isdir(root):
            for dirname, _, filenames in os.walk(root):
                if 'mitbih_train.csv' in filenames:
                    return dirname
    raise FileNotFoundError('mitbih_train.csv not found. Add the "ECG Heartbeat Categorization Dataset" to the notebook.')

DATA_DIR = find_data_dir()
train_df = pd.read_csv(os.path.join(DATA_DIR, 'mitbih_train.csv'), header=None)
test_df  = pd.read_csv(os.path.join(DATA_DIR, 'mitbih_test.csv'),  header=None)

X_full = train_df.iloc[:, :-1].values.astype('float32')
y_full = train_df.iloc[:, -1].values.astype('int64')
X_test = test_df.iloc[:, :-1].values.astype('float32')
y_test = test_df.iloc[:, -1].values.astype('int64')

N_SAMPLES = X_full.shape[1]
t = np.arange(N_SAMPLES) / FS
print('Data folder :', DATA_DIR)
print('Train file  :', X_full.shape, '| Test file:', X_test.shape)
print('Train beats per class:', dict(zip(CLASS_NAMES, np.bincount(y_full, minlength=5))))
print('Test beats per class :', dict(zip(CLASS_NAMES, np.bincount(y_test, minlength=5))))
```

## 3. Split and class weights
Same split as the baseline: 20% of the training file is the validation set (stratified, seed 42). The test file is kept aside.

Two weight schemes are prepared:
- **balanced** (used by the baseline): `w = N / (K * n_c)`. Rare classes get very large weights.
- **sqrt** (change E3): the square root of the balanced weights. Rare classes still count more, but far less extremely.

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_full, y_full, test_size=0.20, stratify=y_full, random_state=SEED)

balanced = compute_class_weight('balanced', classes=np.arange(5), y=y_train)
CLASS_WEIGHTS = {'balanced': balanced, 'sqrt': np.sqrt(balanced)}

print('Train:', X_train.shape, '| Val:', X_val.shape, '| Test:', X_test.shape)
print(pd.DataFrame({'Class': CLASS_NAMES,
                    'Train beats': np.bincount(y_train, minlength=5),
                    'balanced weight': balanced.round(2),
                    'sqrt weight': np.sqrt(balanced).round(2)}).to_string(index=False))
print('Weight ratio F : N  ->  balanced %.0f : 1   |   sqrt %.1f : 1'
      % (balanced[3] / balanced[0], np.sqrt(balanced[3] / balanced[0])))
```

## 4. Light augmentation (change E4)
Each epoch, every **training** beat gets a small random change, so the model never sees exactly the same beat twice:
- amplitude scaled by a factor between 0.9 and 1.1,
- small noise added (standard deviation 0.01, the old Kaggle notebook used 0.5, which buried the beat),
- shifted left or right by up to 3 samples (24 ms).

The zero padding at the end of a beat is left untouched. Validation and test beats are never augmented.

```python
def augment(X, rng):
    # X: (beats, samples). Returns a randomly changed copy.
    n = X.shape[0]
    out = X * rng.uniform(0.9, 1.1, size=(n, 1)).astype('float32')
    out = out + rng.normal(0, 0.01, size=X.shape).astype('float32') * (X > 0)
    shifts = rng.randint(-3, 4, size=n)
    res = out.copy()
    for s in range(-3, 4):
        idx = np.where(shifts == s)[0]
        if len(idx) == 0 or s == 0:
            continue
        if s > 0:      # shift right: repeat the first value at the start
            res[idx, s:] = out[idx, :-s]
            res[idx, :s] = out[idx, :1]
        else:          # shift left: zeros at the end, like the padding
            res[idx, :s] = out[idx, -s:]
            res[idx, s:] = 0.0
    return res.astype('float32')

demo_rng = np.random.RandomState(0)
fig, axes = plt.subplots(1, 3, figsize=(15, 3), sharey=True)
for ax, k in zip(axes, [0, 1, 3]):
    beat = X_train[np.where(y_train == k)[0][0]][None, :]
    ax.plot(t, beat[0], color='black', linewidth=1.8, label='Original')
    for _ in range(3):
        ax.plot(t, augment(beat, demo_rng)[0], linewidth=0.9, alpha=0.8)
    ax.set_title(f'{CLASS_NAMES[k]} beat: original (black) and 3 augmented copies', fontsize=10)
    ax.set_xlabel('Time (s)'); ax.grid(alpha=0.3)
axes[0].set_ylabel('Normalised amplitude')
plt.tight_layout(); savefig('01_augmentation_examples.png'); plt.show()
```

## 5. The two networks
- **cnn**: the baseline network, unchanged (3 convolution blocks, 32 / 64 / 128 filters).
- **resnet** (change E5): 3 residual blocks. Each block has two convolutions and a *shortcut* that adds the block's input to its output. The shortcut lets a deeper network train without the signal fading on the way.

In both networks the last activation before the final pooling is named `gradcam_target`, so Grad-CAM works on either one.

**Batch-norm momentum (part of change E2):** batch normalisation keeps a running average of the signal statistics, and uses that average when predicting. With the default momentum of 0.99 the average updates slowly and can lag behind the changing weights, so the model can do well during training and badly on validation in the same epoch. That pattern is visible in the baseline log. A momentum of 0.9 makes the average follow the weights more closely.

```python
def build_cnn(n_samples, n_classes=5, bn_momentum=0.99):
    inp = keras.Input(shape=(n_samples, 1), name='beat')
    x = inp
    for i, (filters, kernel) in enumerate([(32, 7), (64, 5), (128, 3)], start=1):
        x = layers.Conv1D(filters, kernel, padding='same', name=f'conv{i}')(x)
        x = layers.BatchNormalization(momentum=bn_momentum, name=f'bn{i}')(x)
        x = layers.Activation('relu', name='gradcam_target' if i == 3 else f'relu{i}')(x)
        x = layers.MaxPooling1D(2, name=f'pool{i}')(x)
    x = layers.GlobalAveragePooling1D(name='gap')(x)
    x = layers.Dense(64, activation='relu', name='dense')(x)
    x = layers.Dropout(0.3, name='dropout')(x)
    out = layers.Dense(n_classes, activation='softmax', name='class_probs')(x)
    return keras.Model(inp, out, name='baseline_1d_cnn')

def build_resnet(n_samples, n_classes=5, bn_momentum=0.99):
    inp = keras.Input(shape=(n_samples, 1), name='beat')
    x = layers.Conv1D(32, 7, padding='same', name='stem_conv')(inp)
    x = layers.BatchNormalization(momentum=bn_momentum, name='stem_bn')(x)
    x = layers.Activation('relu', name='stem_relu')(x)
    for i, filters in enumerate([32, 64, 128], start=1):
        shortcut = x
        if int(shortcut.shape[-1]) != filters:      # match the number of channels
            shortcut = layers.Conv1D(filters, 1, padding='same', name=f'res{i}_shortcut')(shortcut)
        y = layers.Conv1D(filters, 5, padding='same', name=f'res{i}_conv_a')(x)
        y = layers.BatchNormalization(momentum=bn_momentum, name=f'res{i}_bn_a')(y)
        y = layers.Activation('relu', name=f'res{i}_relu_a')(y)
        y = layers.Conv1D(filters, 5, padding='same', name=f'res{i}_conv_b')(y)
        y = layers.BatchNormalization(momentum=bn_momentum, name=f'res{i}_bn_b')(y)
        x = layers.Add(name=f'res{i}_add')([shortcut, y])
        x = layers.Activation('relu', name='gradcam_target' if i == 3 else f'res{i}_relu_out')(x)
        x = layers.MaxPooling1D(2, name=f'res{i}_pool')(x)
    x = layers.GlobalAveragePooling1D(name='gap')(x)
    x = layers.Dense(64, activation='relu', name='dense')(x)
    x = layers.Dropout(0.3, name='dropout')(x)
    out = layers.Dense(n_classes, activation='softmax', name='class_probs')(x)
    return keras.Model(inp, out, name='residual_1d_cnn')

BUILDERS = {'cnn': build_cnn, 'resnet': build_resnet}
for name, fn in BUILDERS.items():
    print(f'{name:7s}: {fn(N_SAMPLES).count_params():,} parameters')
```

## 6. Training loop that selects on validation macro-F1 (change E1)
After every epoch the model predicts the validation set and the **macro-F1** is computed. The weights of the epoch with the best validation macro-F1 are kept. If there is no improvement for 3 epochs the learning rate is halved, and after 8 epochs training stops.

Macro-F1 is the average of the five per-class F1 scores, so the rare classes S and F count as much as N. This is the project's main metric, so it is also the one used to choose the model.

```python
def macro_f1(y_true, y_hat):
    return f1_score(y_true, y_hat, labels=np.arange(5), average='macro')

def set_lr(model, lr):
    try:
        model.optimizer.learning_rate.assign(lr)
    except Exception:
        keras.backend.set_value(model.optimizer.lr, lr)

def test_scores(y_true, y_hat):
    p, r, f, s = precision_recall_fscore_support(y_true, y_hat, labels=np.arange(5))
    return {'acc': accuracy_score(y_true, y_hat), 'macro_f1': f.mean(),
            'prec': p, 'rec': r, 'f1': f, 'support': s}

def run_experiment(cfg):
    reset_seeds()
    keras.backend.clear_session()
    model = BUILDERS[cfg['arch']](N_SAMPLES, bn_momentum=cfg['bn_momentum'])
    opt_args = {'clipnorm': cfg['clipnorm']} if cfg['clipnorm'] else {}
    model.compile(optimizer=keras.optimizers.Adam(cfg['lr'], **opt_args),
                  loss='sparse_categorical_crossentropy')
    sample_w = CLASS_WEIGHTS[cfg['weights']][y_train].astype('float32')
    rng = np.random.RandomState(SEED)

    lr, best_f1, best_epoch, best_weights, wait = cfg['lr'], -1.0, 0, None, 0
    log = {'loss': [], 'val_f1': [], 'val_acc': []}
    start = time.time()
    for epoch in range(1, EPOCHS + 1):
        X_epoch = augment(X_train, rng) if cfg['augment'] else X_train
        hist = model.fit(X_epoch[..., None], y_train, sample_weight=sample_w,
                         batch_size=BATCH_SIZE, epochs=1, shuffle=True, verbose=0)
        val_hat = model.predict(X_val[..., None], batch_size=1024, verbose=0).argmax(axis=1)
        vf, va = macro_f1(y_val, val_hat), accuracy_score(y_val, val_hat)
        log['loss'].append(float(hist.history['loss'][0])); log['val_f1'].append(vf); log['val_acc'].append(va)
        improved = vf > best_f1
        print(f"  epoch {epoch:2d}  train loss {log['loss'][-1]:.4f}  val acc {va:.4f}  "
              f"val macro-F1 {vf:.4f}  lr {lr:.5f}{'  <- best' if improved else ''}")
        if improved:
            best_f1, best_epoch, best_weights, wait = vf, epoch, model.get_weights(), 0
        else:
            wait += 1
            if wait % 3 == 0 and lr > 1e-5:
                lr = max(lr * 0.5, 1e-5); set_lr(model, lr)
            if wait >= PATIENCE:
                print(f'  stopped early at epoch {epoch}')
                break
    model.set_weights(best_weights)
    probs_val  = model.predict(X_val[..., None],  batch_size=1024, verbose=0)
    probs_test = model.predict(X_test[..., None], batch_size=1024, verbose=0)
    return {'cfg': cfg, 'log': log, 'weights': best_weights, 'best_epoch': best_epoch,
            'epochs_run': len(log['loss']), 'val_f1': best_f1, 'seconds': time.time() - start,
            'probs_val': probs_val, 'probs_test': probs_test,
            'test': test_scores(y_test, probs_test.argmax(axis=1))}
```

## 7. Run the experiments
Each experiment keeps everything from the one before it and adds one change.

```python
EXPERIMENTS = [
    {'id': 'E1', 'change': 'Best epoch chosen by validation macro-F1',
     'arch': 'cnn', 'lr': 1e-3, 'clipnorm': None, 'bn_momentum': 0.99, 'weights': 'balanced', 'augment': False},
    {'id': 'E2', 'change': '+ stable training (lr 0.0005, clipping, BN momentum 0.9)',
     'arch': 'cnn', 'lr': 5e-4, 'clipnorm': 1.0, 'bn_momentum': 0.9, 'weights': 'balanced', 'augment': False},
    {'id': 'E3', 'change': '+ softer (square-root) class weights',
     'arch': 'cnn', 'lr': 5e-4, 'clipnorm': 1.0, 'bn_momentum': 0.9, 'weights': 'sqrt', 'augment': False},
    {'id': 'E4', 'change': '+ light augmentation',
     'arch': 'cnn', 'lr': 5e-4, 'clipnorm': 1.0, 'bn_momentum': 0.9, 'weights': 'sqrt', 'augment': True},
    {'id': 'E5', 'change': '+ deeper residual network',
     'arch': 'resnet', 'lr': 5e-4, 'clipnorm': 1.0, 'bn_momentum': 0.9, 'weights': 'sqrt', 'augment': True},
]

runs = {}
total_start = time.time()
for cfg in EXPERIMENTS:
    if cfg['id'] not in RUN_IDS:
        continue
    print(f"=== {cfg['id']}: {cfg['change']} ===")
    r = runs[cfg['id']] = run_experiment(cfg)
    print(f"  best epoch {r['best_epoch']} of {r['epochs_run']}  |  val macro-F1 {100 * r['val_f1']:.2f}%  |  "
          f"test accuracy {100 * r['test']['acc']:.2f}%  test macro-F1 {100 * r['test']['macro_f1']:.2f}%  |  "
          f"{r['seconds']:.0f} s")
    print()
print(f'Total training time: {(time.time() - total_start) / 60:.1f} min')
```

### How stable was each training run?
In the baseline, validation accuracy swung between 14% and 95% from one epoch to the next. A smoother line here means the change made training more stable.

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 4))
for eid, r in runs.items():
    ep = np.arange(1, r['epochs_run'] + 1)
    ax[0].plot(ep, r['log']['val_f1'], marker='o', markersize=3, label=eid)
    ax[1].plot(ep, r['log']['val_acc'], marker='o', markersize=3, label=eid)
ax[0].set_title('Validation macro-F1 per epoch'); ax[1].set_title('Validation accuracy per epoch')
for a in ax:
    a.set_xlabel('Epoch'); a.set_ylim(0, 1.02); a.grid(alpha=0.3); a.legend()
plt.tight_layout(); savefig('02_validation_curves.png'); plt.show()
```

## 8. Choose the final model (on validation only) and tune the decision thresholds
**Choosing the model:** the experiment with the highest *validation* macro-F1 becomes the final model. Test scores are not looked at for this choice.

**Threshold tuning:** normally the model picks the class with the highest probability. Here each class probability is first multiplied by a scale factor. A factor below 1 means the model must be more sure before it chooses that class, which reduces false alarms for that class. The five factors are searched on the validation set to maximise validation macro-F1, then applied once to the test set. No retraining is needed.

```python
best_id = max(runs, key=lambda k: runs[k]['val_f1'])
best = runs[best_id]
print(f"Final model: {best_id} ({best['cfg']['change']}), validation macro-F1 {100 * best['val_f1']:.2f}%")

def tune_scales(probs, y_true, grid=(0.05, 0.1, 0.2, 0.3, 0.5, 0.7, 1.0, 1.5, 2.0, 3.0, 5.0), passes=2):
    scales = np.ones(probs.shape[1])
    best_score = macro_f1(y_true, probs.argmax(axis=1))
    for _ in range(passes):
        for k in range(probs.shape[1]):
            for g in grid:
                trial = scales.copy(); trial[k] = g
                score = macro_f1(y_true, (probs * trial).argmax(axis=1))
                if score > best_score + 1e-6:
                    best_score, scales = score, trial
    return scales, best_score

scales, val_f1_tuned = tune_scales(best['probs_val'], y_val)
print('Scale factor per class:', {c: float(v) for c, v in zip(CLASS_NAMES, scales)})
print(f"Validation macro-F1: {100 * best['val_f1']:.2f}% before tuning -> {100 * val_f1_tuned:.2f}% after "
      '(the "after" value is optimistic, because the factors were fitted on this same set)')

probs = best['probs_test']
y_pred_raw = probs.argmax(axis=1)                 # final model, plain argmax
y_pred     = (probs * scales).argmax(axis=1)      # final model + tuned thresholds
final_raw, final = test_scores(y_test, y_pred_raw), test_scores(y_test, y_pred)
```

```python
def pct(x):
    return round(100 * x, 2)

rows = [{'ID': 'B0', 'Change': 'Baseline (earlier run)', 'Val macro-F1': np.nan,
         'Test acc': pct(BASELINE['acc']), 'Test macro-F1': pct(BASELINE['macro_f1']),
         'Recall S': pct(BASELINE['rec_S']), 'Recall F': pct(BASELINE['rec_F']),
         'Precision S': pct(BASELINE['prec_S']), 'Precision F': pct(BASELINE['prec_F']), 'Best epoch': 11}]
for eid, r in runs.items():
    s = r['test']
    rows.append({'ID': eid, 'Change': r['cfg']['change'], 'Val macro-F1': pct(r['val_f1']),
                 'Test acc': pct(s['acc']), 'Test macro-F1': pct(s['macro_f1']),
                 'Recall S': pct(s['rec'][1]), 'Recall F': pct(s['rec'][3]),
                 'Precision S': pct(s['prec'][1]), 'Precision F': pct(s['prec'][3]),
                 'Best epoch': r['best_epoch']})
rows.append({'ID': 'Final', 'Change': f'{best_id} + tuned thresholds', 'Val macro-F1': pct(val_f1_tuned),
             'Test acc': pct(final['acc']), 'Test macro-F1': pct(final['macro_f1']),
             'Recall S': pct(final['rec'][1]), 'Recall F': pct(final['rec'][3]),
             'Precision S': pct(final['prec'][1]), 'Precision F': pct(final['prec'][3]),
             'Best epoch': best['best_epoch']})
table = pd.DataFrame(rows)
table.to_csv('experiment_results.csv', index=False)
pd.set_option('display.width', 250); pd.set_option('display.max_colwidth', 60)
print(table.to_string(index=False))

x = np.arange(len(table)); w = 0.27
plt.figure(figsize=(11, 4.2))
plt.bar(x - w, table['Test macro-F1'], w, label='Macro-F1', color='#1f3b5c')
plt.bar(x,     table['Recall S'],      w, label='Recall S', color='#3b6ea5')
plt.bar(x + w, table['Recall F'],      w, label='Recall F', color='#9bb7d4')
plt.xticks(x, table['ID']); plt.ylim(0, 105); plt.ylabel('Score on the test set (%)')
plt.title('Effect of each change (B0 = baseline, Final = best model + tuned thresholds)')
plt.legend(loc='lower right'); plt.grid(axis='y', alpha=0.3)
savefig('03_experiment_comparison.png'); plt.show()
```

## 11. Grad-CAM on the final model
Same method as in the baseline notebook, applied to the layer named `gradcam_target`. Darker red = more important for the decision. This is still a qualitative preview.

```python
keras.backend.clear_session()
model = BUILDERS[best['cfg']['arch']](N_SAMPLES, bn_momentum=best['cfg']['bn_momentum'])
model.set_weights(best['weights'])
model.save('improved_1d_cnn.h5')

def grad_cam(model, beat, class_idx, layer_name='gradcam_target'):
    grad_model = keras.Model(model.inputs, [model.get_layer(layer_name).output, model.output])
    x = tf.convert_to_tensor(beat.reshape(1, -1, 1), dtype=tf.float32)
    with tf.GradientTape() as tape:
        conv_out, preds = grad_model(x, training=False)
        if isinstance(preds, (list, tuple)):
            preds = preds[0]
        score = preds[:, class_idx]
    grads = tape.gradient(score, conv_out)                 # (1, steps, channels)
    alpha = tf.reduce_mean(grads, axis=1, keepdims=True)   # one weight per channel
    cam = tf.nn.relu(tf.reduce_sum(alpha * conv_out, axis=-1))[0].numpy()
    if cam.max() > 0:
        cam = cam / cam.max()
    steps = np.linspace(0, len(beat) - 1, len(cam))
    return np.interp(np.arange(len(beat)), steps, cam)

im = None
fig, axes = plt.subplots(1, 5, figsize=(20, 3.4), sharey=True)
for k, ax in enumerate(axes):
    correct = np.where((y_test == k) & (y_pred_raw == k))[0]
    if len(correct) == 0:
        ax.set_title(f'{CLASS_NAMES[k]}: no correctly classified beat'); ax.axis('off'); continue
    idx = correct[np.argmax(probs[correct, k])]            # the most confident correct beat
    beat = X_test[idx]
    cam = grad_cam(model, beat, k)
    im = ax.imshow(cam[None, :], aspect='auto', cmap='Reds', vmin=0, vmax=1, alpha=0.75,
                   extent=[0, N_SAMPLES / FS, beat.min() - 0.05, beat.max() + 0.05])
    ax.plot(t, beat, color='black', linewidth=1.6)
    ax.set_title(f'True {CLASS_NAMES[k]}, predicted {CLASS_NAMES[k]} (p = {probs[idx, k]:.2f})', fontsize=10)
    ax.set_xlabel('Time (s)')
axes[0].set_ylabel('Normalised amplitude')
if im is not None:
    fig.colorbar(im, ax=axes, fraction=0.012, pad=0.01, label='Grad-CAM importance')
savefig('06_gradcam_final.png'); plt.show()
```

```python
# ============================================================
# DOWNLOAD EVERYTHING AS ONE ZIP
# ============================================================

import os
import shutil

ZIP_NAME = "ECG_Arrhythmia_Complete_Results"
RESULTS_DIR = ZIP_NAME

# Remove old package if it exists
if os.path.exists(RESULTS_DIR):
    shutil.rmtree(RESULTS_DIR)

os.makedirs(RESULTS_DIR, exist_ok=True)


# ------------------------------------------------------------
# 1. COPY ALL FIGURES
# ------------------------------------------------------------

if os.path.exists("figures"):

    shutil.copytree(
        "figures",
        os.path.join(RESULTS_DIR, "figures"),
        dirs_exist_ok=True
    )


# ------------------------------------------------------------
# 2. COPY ALL CSV / RESULT FILES
# ------------------------------------------------------------

for root, dirs, files in os.walk("."):

    # Don't copy the results folder into itself
    if RESULTS_DIR in root:
        continue

    for file in files:

        if file.endswith((".csv", ".xlsx", ".json", ".txt")):

            source = os.path.join(root, file)

            # Skip Kaggle/system files
            if "/kaggle/" in source.replace("\\", "/"):
                continue

            try:
                shutil.copy2(
                    source,
                    os.path.join(RESULTS_DIR, file)
                )
            except:
                pass


# ------------------------------------------------------------
# 3. COPY TRAINED MODELS
# ------------------------------------------------------------

model_extensions = (
    ".h5",
    ".keras",
    ".weights.h5"
)

for root, dirs, files in os.walk("."):

    if RESULTS_DIR in root:
        continue

    for file in files:

        if file.endswith(model_extensions):

            source = os.path.join(root, file)

            try:
                shutil.copy2(
                    source,
                    os.path.join(RESULTS_DIR, file)
                )
            except:
                pass


# ------------------------------------------------------------
# 4. COPY NOTEBOOK
# ------------------------------------------------------------

# If your notebook has this name, copy it if available
for root, dirs, files in os.walk("/kaggle/working"):

    for file in files:

        if file.endswith(".ipynb"):

            source = os.path.join(root, file)

            try:
                shutil.copy2(
                    source,
                    os.path.join(RESULTS_DIR, file)
                )
            except:
                pass


# ------------------------------------------------------------
# 5. CREATE ZIP
# ------------------------------------------------------------

zip_path = shutil.make_archive(
    ZIP_NAME,
    "zip",
    RESULTS_DIR
)


# ------------------------------------------------------------
# 6. SHOW CONTENTS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ECG RESULTS PACKAGE CREATED")
print("=" * 70)

print(f"\nZIP: {zip_path}")

print("\nContents:")

for root, dirs, files in os.walk(RESULTS_DIR):

    for file in files:

        path = os.path.join(root, file)
        print("  ✓", path)


print("\n" + "=" * 70)
print("DOWNLOAD THIS FILE FROM KAGGLE:")
print(f"{ZIP_NAME}.zip")
print("=" * 70)
```
